```python
# Import required libraries
import re

import nltk #Natural Language Toolkit.

from nltk.tokenize import word_tokenize, sent_tokenize
# word_tokenize → sentence ko individual words/tokens mein todta hai.
# sent_tokenize → paragraph ko individual sentences mein todta hai.

from nltk.corpus import stopwords
# stopwords common words ki list provide karta hai jo NLP mein often remove kiye jaate hain.

from nltk.stem import PorterStemmer
# PorterStemmer words ko unke root/stem form mein convert karta hai.

from nltk.util import ngrams
# ngrams ka use word combinations banane ke liye hota hai.

import spacy
# ➡️ spaCy ek powerful NLP library hai.

# Isse hum:

# tokenization
# POS tagging
# Named Entity Recognition (NER)
# dependency parsing
# lemmatization

# jaise tasks kar sakte hain.
```

```python
# Download NLTK resources
nltk.download('punkt_tab')

# NLTK ka Punkt tokenizer data download karta hai.
# Ye mainly word_tokenize() aur sent_tokenize() jaise tokenization functions ke liye required data provide karta hai.

nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger') # ➡️ NLTK ka POS Tagger data download karta hai. | POS = Part of Speech
nltk.download('averaged_perceptron_tagger_eng') # ➡️ English language ke liye POS tagging model/data download karta hai.
```

```python
# Load spaCy model
nlp = spacy.load('en_core_web_sm')
# Sample text
text = "Artificial Intelligence is everywhere! In 2026, millions of people use AI every day. It is powerful, useful, and exciting!!!"
```

```python
# Step 1: Data Cleaning
def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r'\d+', '', text)  # Remove numbers
    text = re.sub(r'[^\w\s]', '', text)  # Remove punctuation
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text
cleaned_text = clean_text(text)
print("Cleaned Text:", cleaned_text)
```

```python
# Step 2: Tokenization
nltk_words = word_tokenize(cleaned_text)  # NLTK word tokenization
nltk_sentences = sent_tokenize(text)  # NLTK sentence tokenization
print("\nNLTK Word Tokens:", nltk_words)
print("\nNLTK Sentence Tokens:", nltk_sentences)
# SpaCy Tokenization
doc = nlp(cleaned_text)
spacy_tokens = [token.text for token in doc]
print("\nSpaCy Tokens:", spacy_tokens)
```

```python
# Step 3: Stemming (NLTK)
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in nltk_words]
print("\nStemmed Words (NLTK):", stemmed_words)
```

```python
# Step 4: Lemmatization (spaCy)
lemmatized_words = [token.lemma_ for token in doc]
print("\nLemmatized Words (spaCy):", lemmatized_words)
```

```python
# Step 5: Stop Words Removal
stop_words = set(stopwords.words('english'))
filtered_words_nltk = [word for word in nltk_words if word not in stop_words]
print("\nFiltered Words (NLTK):", filtered_words_nltk)
filtered_words_spacy = [token.text for token in doc if not token.is_stop]
print("\nFiltered Words (spaCy):", filtered_words_spacy)
```

```python
# Step 6: POS Tagging
nltk_pos_tags = nltk.pos_tag(nltk_words)
print("\nPOS Tags (NLTK):", nltk_pos_tags)
spacy_pos_tags = [(token.text, token.pos_) for token in doc]
print("\nPOS Tags (spaCy):", spacy_pos_tags)
```

```python
# Step 7: N-Gram Modelling
# Trigram with NLTK
trigrams = list(ngrams(nltk_words, 3))
print("\nTrigrams (NLTK):", trigrams)
# Word Gram Modelling with NLTK
word_grams = list(ngrams(filtered_words_nltk, 2))  # Bigram example
print("\nWord Grams (NLTK):", word_grams)
```
