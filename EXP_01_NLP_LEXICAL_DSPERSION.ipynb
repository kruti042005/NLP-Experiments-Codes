WITH NLTK

```python
import nltk
from nltk.corpus import gutenberg
from nltk import word_tokenize
import matplotlib.pyplot as plt

# Ensure required NLTK packages are downloaded
nltk.download('gutenberg')
nltk.download('punkt_tab')

# Load a sample corpus (Moby Dick by Herman Melville)
text = gutenberg.raw('melville-moby_dick.txt')

# Tokenize the text into words
tokens = word_tokenize(text)

# Define the target words to be analyzed
target_words = ["whale", "ship", "sea", "captain"]

# Function to generate lexical dispersion data
def lexical_dispersion_plot(tokens, target_words):
    # Identify the positions of each target word in the tokenized text
    word_positions = {word: [] for word in target_words}
    for idx, word in enumerate(tokens):
        if word.lower() in word_positions:
            word_positions[word.lower()].append(idx)

    # Plot the dispersion plot
    plt.figure(figsize=(12, 6))
    for i, word in enumerate(target_words):
        y = [i] * len(word_positions[word])  # Set a fixed y-value for each word
        plt.plot(word_positions[word], y, "|", label=word)

    plt.yticks(range(len(target_words)), target_words)
    plt.ylim(-1, len(target_words))
    plt.xlabel("Word Position")
    plt.title("Lexical Dispersion Plot")
    plt.legend()
    plt.show()

# Call the function to display the lexical dispersion plot
lexical_dispersion_plot(tokens, target_words)
```

WITH SpaCY

```python
import spacy
import matplotlib.pyplot as plt

# Load SpaCy's English language model
nlp = spacy.load("en_core_web_sm")

# Define a sample text (or load from a file if desired)
text = """In the digital age, technology has changed the way people communicate, learn, and work. Modern technology allows students to access information quickly and communicate with teachers through online platforms. However, effective communication is still important because technology cannot completely replace human interaction. Good communication helps people share ideas, solve problems, and build relationships. As education becomes increasingly digital, students use technology to attend classes, read books, and collaborate with others. Digital education also provides opportunities for people who cannot attend traditional classrooms. Therefore, technology, communication, and education continue to influence the way society learns and connects."""

# Process the text with SpaCy
doc = nlp(text)

# Define target words for lexical dispersion
target_words = ["technology" , "communication" , "education" , "students"]

# Find the positions of target words in the text
def lexical_dispersion_plot(doc, target_words):
    # Dictionary to store positions of target words
    word_positions = {word: [] for word in target_words}

    # Populate positions for each target word
    for idx, token in enumerate(doc):
        if token.text.lower() in word_positions:
            word_positions[token.text.lower()].append(idx)

    # Plot the dispersion plot using Matplotlib
    plt.figure(figsize=(12, 6))
    for i, word in enumerate(target_words):
        y = [i] * len(word_positions[word])  # Fixed y-position for each word
        plt.plot(word_positions[word], y, '|', label=word)  # Scatter plot for each word
    plt.yticks(range(len(target_words)), target_words)
    plt.ylim(-1, len(target_words))
    plt.xlabel("Word Position")
    plt.title("Lexical Dispersion Plot using SpaCy")
    plt.legend()
    plt.show()

# Call the function to display the lexical dispersion plot
lexical_dispersion_plot(doc, target_words)
```
